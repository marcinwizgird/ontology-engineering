## The chapter's central claim

> **Consistency is not correctness.**

Chapters 3 and 4 gave you a reasoner. A reasoner tells you whether your axioms *can* all be true together. It cannot tell you whether they say what you meant. `Person ⊑ Student` is perfectly consistent and completely wrong, and §5.2 is about the second kind of error.
